## 1.1 Data Preprocessing

```python
# run only one time
movies = movies.merge(credits, on='title')
```

```python
movies.head()
```

```python
movies.info()
```

```python
feature = ['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']

movies = movies[feature]
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
movies.iloc[0]
```

### Genres

```python
movies['genres'][0]
```

```python
# '[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]'
# ['Action', 'Adventure', 'Fantasy', 'SciFi']
```

```python
import ast
```

```python
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])

    return L
```

```python
convert(movies['genres'][0])
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies.head()
```

### Keywords

```python
movies.iloc[0].keywords
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
movies.head()
```

### Cast

```python
movies.iloc[0].cast
```

```python
def convert_cast(obj):
    L = []
    cnt = 0
    for i in ast.literal_eval(obj):
        if(cnt >= 4):
            break
        L.append(i['name'])
        cnt+=1
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert_cast)
movies.head()
```

### Crew

```python
movies.iloc[0].crew
```

```python
def convert_crew(obj):
    L = []

    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break

    return L
```

```python
movies['crew'] = movies['crew'].apply(convert_crew)
```

```python
movies.head()
```

### Overview

```python
movies['overview'] = movies['overview'].apply(lambda x : x.split())
```

### Removing space

```python
movies['genres']= movies['genres'].apply(lambda x : [i.replace(" ", "") for i in x])
movies['keywords']= movies['keywords'].apply(lambda x : [i.replace(" ", "") for i in x])
movies['cast']= movies['cast'].apply(lambda x : [i.replace(" ", "") for i in x])
movies['crew']= movies['crew'].apply(lambda x : [i.replace(" ", "") for i in x])
```

```python
movies.head()
```

### Joining

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies.head()
```

### New DF & some changes

```python
df = movies[['movie_id', 'title', 'tags']]
```

```python
df['tags'] = df['tags'].apply(lambda x : " ".join(x))
```

```python
df.head()
```

```python
df['tags'][0]
```

```python
df['tags'] = df['tags'].apply(lambda x : x.lower())
```

```python
df.head()
```

### Removing Similler Word

```python
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
def stem(text):
    L = []

    for i in text.split():
        L.append(ps.stem(i))
    
    return " ".join(L)
```

```python
df['tags'] = df['tags'].apply(stem)
```

```python
df.head()
```

## 1.2 Vectorization

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(df['tags']).toarray()
```

```python
vectors
```

### Vector Analysis

```python
cv.get_feature_names_out()
```

### Consine Similarity

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
similarity
```

## 2.3 Sorting the distance

similarity list has similarity score between two movies
<br>
similarity[0] has score of similarity between all the movies with respect to the movie on the 0th index 
<br>
so if we sort the list we can't say like movie at the 0th index has higher similarity with movie at 8th index
<br>

```python
sorted(similarity[0], reverse=True)
```

to solve above issue we will use enumarate

## 2.4 Enumeration

```python
# will give enumerate object
enumerate(similarity[0])
```

```python
# by converting the enumerate object into list we will have, what we need (index, simillarity_score)

list(enumerate(similarity[0]))
```

```python
# sorting the list
sorted(enumerate(similarity[0]), reverse=True, key=lambda x : x[1])
```

```python
# getting top 5 similer movies
sorted(enumerate(similarity[0]), reverse=True, key=lambda x : x[1])[1:6]
```

## Function by combining all of the above mentioned steps

```python
def recommend_movies(movie):
    movie_index = df[df['title'] == movie].index[0]

    distance_with_all = similarity[movie_index]
    movies_list = sorted(list(enumerate(distance_with_all)), reverse=True, key=lambda x : x[1])[1:6]

    
    for i in movies_list:
        print(df.iloc[i[0]].title)
```

```python
recommend_movies('Batman Begins')
```
